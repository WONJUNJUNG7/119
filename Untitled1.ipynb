## Calculate Quantiles for Fire Risk Index

### Subtask:
Calculate the 25th, 50th (median), and 75th percentiles for the '화재위험지수' to define thresholds for classifying districts.


**Reasoning**:
I need to calculate the 25th, 50th, and 75th percentiles of the '화재위험지수' column in `merged_df` as instructed. I will use the `.quantile()` method for this, store the results in variables, and then print them.



```python
q1 = merged_df['화재위험지수'].quantile(0.25)
q2 = merged_df['화재위험지수'].quantile(0.50)
q3 = merged_df['화재위험지수'].quantile(0.75)

print(f"25th percentile (Q1) of 화재위험지수: {q1:.2f}")
print(f"50th percentile (Median) of 화재위험지수: {q2:.2f}")
print(f"75th percentile (Q3) of 화재위험지수: {q3:.2f}")
```

## Classify Districts

### Subtask:
Categorize each administrative district into 'Danger Zone' (high '화재위험지수'), 'Safe Zone' (low '화재위험지수'), or 'Moderate Zone' based on the calculated quantiles.


**Reasoning**:
As per the instructions, I need to classify the districts into 'Danger Zone', 'Safe Zone', or 'Moderate Zone' based on the '화재위험지수' and the previously calculated quantiles (q1, q2, q3). I will use `np.select` for efficient conditional assignment to a new 'Risk_Zone' column.



```python
import numpy as np

conditions = [
    merged_df['화재위험지수'] > q3,
    merged_df['화재위험지수'] <= q1
]
choices = ['Danger Zone', 'Safe Zone']

merged_df['Risk_Zone'] = np.select(conditions, choices, default='Moderate Zone')

print("Merged DataFrame with 'Risk_Zone' column:")
display(merged_df.head())
```
